```python
from keras.preprocessing import image
#from keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import os
import shutil
import numpy as np

# 1. 
img_dir = 'C:/Users/Administrator/AI/Mask_std/origin'
save_dir = 'C:/Users/Administrator/AI/Mask_std/aug'
word = ['0', '1']

# 2. 
for key in word:
    os.makedirs(os.path.join(img_dir, key), exist_ok=True)
    os.makedirs(os.path.join(save_dir, key), exist_ok=True)

# 3. 
all_files = [f for f in os.listdir(img_dir) if f.lower().endswith('.jpg')]
for file in all_files:
    if '-' in file:
        label = file.split('-')[0]
        if label in word:
            src = os.path.join(img_dir, file)
            dst = os.path.join(img_dir, label, file)
            try:
                shutil.move(src, dst)
            except:
                pass

# 4. 
trainDataGen = ImageDataGenerator(
    rescale=1./255,
    zoom_range=[1.0, 1.0],
    brightness_range=[1.0, 1.0],
    height_shift_range=0.2,
    width_shift_range=0.15,
    fill_mode='nearest'
)

# 5. 
for img_name in word:
    target_origin_dir = os.path.join(img_dir, img_name)
    target_save_dir = os.path.join(save_dir, img_name)
    
    files_in_folder = [f for f in os.listdir(target_origin_dir) 
                       if f.lower().endswith((".jpg", ".jpeg", ".png"))]
    
    print(f"Foler '{img_name}': {len(files_in_folder)} of images.")

    for file_name in files_in_folder:
        img_path = os.path.join(target_origin_dir, file_name)
        
        try:
            img = image.load_img(img_path, target_size=(128, 128))
            img_tensor = image.img_to_array(img)
            img_tensor = np.expand_dims(img_tensor, axis=0)
            
            i = 0
            for batch in trainDataGen.flow(
                img_tensor, 
                batch_size=1, 
                save_to_dir=target_save_dir, 
                save_prefix=f"aug_{img_name}", 
                save_format='png'
            ):
                i += 1
                if i >= 30:
                    break
        except Exception as e:
            print(f"File {file_name} Error: {e}")

    print(f"--- Class {img_name} Done! ---")
```
